```python
 #Write a Python program to find the factorial of a number using a function, display the result, and validate the input.
```

```python
a = int(input("Enter a number: "))
def factorial(a):
    f = 1
    for i in range(1,a + 1):
        f = f * i
    return f

if a >= 0:
    print(factorial(a))
```

```python
num= 0
def fact(x):
    global num
    num+= 1
    if x== 1:
        return 1
    return x* fact(x-1)

print(fact(5))
print(num)
```

```python
# Raise a custom exception
```

```python
class MyError(Exception):
    pass
age = int(input("Enter age: "))
if age < 18:
    raise MyError("Age is too low!")
else:
    print("Allowed")
```

```python
# Handle exception
```

```python
class MyError(Exception):
    pass
try:
    age = int(input("Enter age: "))
    if age < 18:
        raise MyError("Not allowed")
    print("Allowed")
except MyError as e:
    print("Error:", e)
```

```python
# Display user-friendly message
```

```python
class MyError(Exception):
    pass
try:
    age = int(input("Enter age: "))
    if age < 18:
        raise MyError
    print("You are allowed")
except MyError:
    print("Sorry! You are not eligible (must be 18 or above).")
```
