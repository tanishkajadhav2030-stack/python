```python
#**key value
```

```python
def info(**data):
    print(data)

info(name="tanishka",age=17)
```

```python
#*args(multiple value)
```

```python
def add(*num):
    print(sum(num))

add(10,20,30)
```

```python
def mul(*num):
    result=1
    for x in num:
        result*=x
    print(result)

mul(10,20,30)
```
