```python
# 13.Write a Python program to calculate electricity bill charges based on unit slab rates and display the final bill amount.
```

```python
unit=int(input("Enter electricity units consumed"))
bill = 0
if unit<=100:
    bill = unit*1.5
elif unit<= 200:
    bill=(100 * 1.5)+(unit- 100)*2.5
elif unit<= 300:
    bill=(100*1.5) + (100* 2.5)+(unit- 200)*4
else:
    bill = (100*1.5) + (100 * 2.5) + (100 * 4)+(unit-300)*6
print("Total Electricity Bill=₹", bill)
```

```python
# 23.Write a Python program to accept a product price, apply discount based on price range, and calculate the final price.
```

```python
price = float(input("Enter product price"))
print("Product price is", price)
```

```python
#apply discount on price range and final price
```

```python
if price <= 1000:
    discount = 0
elif price <= 5000:
    discount = price * 0.05
elif price <= 10000:
    discount = price * 0.10
else:
    discount = price * 0.15
final_price = price - discount

print("Discount =", discount)
print("Final Price =", final_price)
```
