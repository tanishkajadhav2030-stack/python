```python
cat=["A","B","C"]
values=[10,20,15]
```

```python
plt.bar(cat,values)
plt.xlabel("cat")
plt.ylabel("value")

plt.title("simple bar")
plt.show()
```

```python
plt.barh(cat,values)
```

```python
labels=["A","B","C"]
size=[30,40,30]
```

```python
plt.pie(size,labels=labels,autopct='%1.1f%%')
plt.title("simple pie ")
```

```python
plt.pie(size,labels=labels,autopct='%.2f')
plt.title("simple pie ")
```

```python
plt.pie(size,labels=labels,autopct='%1.0f%%')
plt.title("simple pie ")
```

```python
score=[39,41,43,46,48,56,67,69,75,82]
counts,bins,patches=plt.hist(score,bins=5,edgecolor='black',color='skyblue')
plt.bar_label(patches)
plt.title("histogram of student score")
plt.xlabel("score range")
plt.ylabel("number of student")
plt.show()
```
