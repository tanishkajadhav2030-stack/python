```python
try:
    x=int("abc")
except(ValueError,TypeError):
    print("value error or type error occured")
```

```python
#generic exception
#catches any kind of error when you don't specify the exact type
```

```python
try:
    x=int("abc")
except Exception as e :
    print("error occured:",e)
```

```python
try:
    x=10/0
except Exception as e:
    print("error occured:",e)
```

```python
#else block
#runs only when no error occurs in the try block.
```

```python
try:
    x=10/2
except ZeroDivisionError:
    print("error")
else:
    print("success:",x)
```

```python
#finally block
#runs every time ,whether an erros happens or not ,often used to cleanup
```

```python
try:
    x=10/0
except ZeroDivisionError:
    print("error")
finally:
    print("this always runs")
```
